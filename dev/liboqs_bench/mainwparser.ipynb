```python
#%%writefile __main__bkp.py
'''Main entry point for user interaction.'''

import sys
import logging
logging.captureWarnings(True)
logging.getLogger('oqs.oqs').disabled = True
from oqs import oqs
import argparse
from liboqs_bench import liboqs_bench

def main():
    
    '''
    Main function for calling liboqs_bench package with arguments, e.g. from the command line.
    
    Parses command line arguments, passes them to the liboqs_bench function, and prints the result.
    '''

    alg_choices=['KeyGeneration', 'Encapsulation', 'Decapsulation', 'complete']
    scheme_choices=[scheme_name for scheme_name in oqs.get_enabled_kem_mechanisms()].append('all')
    
    parser = argparse.ArgumentParser(prog='liboqs_bench')
    parser.add_argument('alg',  metavar='algorithm', choices=alg_choices, nargs='?',
                        help=f'the desired algorithm or "complete": {alg_choices}')
    parser.add_argument('scheme', nargs='*',
                        choices=[scheme_name for scheme_name in oqs.get_enabled_kem_mechanisms()].append('all'),
                       help=f'at least one KEM scheme or "all" - for full list use -l option')
    parser.add_argument('-i', '--iterations', type=int, default=1000, metavar='int', help='the number of iterations to run for the benchmark')
    parser.add_argument('-l', '--list-schemes', action='store_true', help='print the full list of supported KEM schemes')
    parser.add_argument('-d', '--debug')
    parser.add_argument('-v', '--version', action='version', version='%(prog)s-0.1.0')
    args = parser.parse_args()

#     Debugging block
#    print(f'alg: {args.alg},scheme: {args.scheme}, iterations: {args.iterations}, list-schemes: {args.list_schemes}')

    if args.list_schemes:
            schemes_str = '\nSupported Schemes:\n'
            for scheme_name in oqs.get_enabled_kem_mechanisms():
                schemes_str += f'\n"{scheme_name}"'
            print(schemes_str)     
    else:

        results = liboqs_bench(args.alg, *args.scheme, iterations=args.iterations)
    
        for outer in results:
            print()
            print(outer[0])
            for result in outer[1].items():
                print(result)
            
if __name__ == '__main__':
    sys.exit(main())
```
